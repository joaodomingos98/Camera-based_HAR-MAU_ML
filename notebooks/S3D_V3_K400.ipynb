## Setting Seed Function

```python
def set_seed(seed=42, deterministic=True):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    # A100 Tensor Cores: TF32 accelerates conv/matmul with negligible
    # accuracy impact. Safe to leave on regardless of determinism.
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

    if deterministic:
        # Fully reproducible, but slower (no cuDNN kernel autotuning).
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    else:
        # Faster: cuDNN autotunes the best kernels for our fixed input sizes
        # (64x224x224). Results are no longer bit-for-bit reproducible.
        torch.backends.cudnn.deterministic = False
        torch.backends.cudnn.benchmark = True

    print(f"Seed set to: {seed} (deterministic={deterministic})")
```

## Data Partitioning & Splitting

```python
def partition_kth_dataset(orig_dir, target_dir="./kth_split", test_split=0.2, val_split=0.2, seed=0):
    """
    Reads 00sequences.txt, extracts active frame clips, and organizes 
    them into train/validation/test folders based on randomly selecting subject IDs.
    
    Parameters:
    - orig_dir: Path to original KTH dataset
    - target_dir: Where to save the partitioned chunks
    - test_split: Percentage of total subjects reserved for testing (default 0.2 = 20%)
    - val_split: Percentage of the *remaining* subjects reserved for validation (default 0.2 = 20%)
    - seed: Random seed for reproducibility
    """
    # Check if already partitioned
    if os.path.exists(target_dir) and os.path.exists(os.path.join(target_dir, "train")):
        print(f"Dataset already partitioned at '{target_dir}'. Skipping partitioning process.")
        print("Note: If you want to re-roll the random splits, you must delete the folder first.")
        return target_dir
    
    # Randomly Split the 25 Subjects
    random.seed(seed)
    
    # Generate list of subjects ['01', '02', ... '25']
    all_subjects = [f"{i:02d}" for i in range(1, 26)]
    random.shuffle(all_subjects)

    # Calculate partition sizes
    num_total = len(all_subjects)
    num_test = int(num_total * test_split)            # 20% of 25 = 5 subjects
    num_train_val = num_total - num_test              # Remaining 20 subjects
    num_val = int(num_train_val * val_split)          # 20% of 20 = 4 subjects
    
    # Assign subjects to partitions
    test_p = all_subjects[:num_test]
    val_p = all_subjects[num_test : num_test + num_val]
    train_p = all_subjects[num_test + num_val :]

    print(f"Partitioning dataset into {target_dir}...")
    print(f"Subject Splits -> Train: {len(train_p)}, Val: {len(val_p)}, Test: {len(test_p)}")
    
    os.makedirs(target_dir, exist_ok=True)
    
    # Save the exact split logic to a file
    with open(os.path.join(target_dir, "split_records.json"), "w") as f:
        json.dump({"train": train_p, "validation": val_p, "test": test_p}, f, indent=4)

    seq_file = os.path.join(orig_dir, "00sequences.txt")
    if not os.path.exists(seq_file):
        print(f"Error: Could not find 00sequences.txt in {orig_dir}!")
        return orig_dir

    # 2. Parse the sequence file
    with open(seq_file, 'r') as f:
        lines = f.readlines()

    for line in lines:
        line = line.strip()
        if not line or "frames" not in line or "*missing*" in line:
            continue

        # Parse line: person01_boxing_d1  frames  1-95, 96-185, ...
        parts = line.split()
        vid_name = parts[0]

        # Extract frame ranges using regex to grab numbers securely
        ranges_str = " ".join(parts[2:])
        ranges = re.findall(r'(\d+)-(\d+)', ranges_str)

        # Determine Class and Split
        person_id = vid_name.split('_')[0].replace('person', '')
        action = vid_name.split('_')[1]

        # Map to the dynamically generated random splits
        if person_id in train_p: split = 'train'
        elif person_id in val_p: split = 'validation'
        elif person_id in test_p: split = 'test'
        else: continue

        # Locate original video
        orig_vid_path = os.path.join(orig_dir, action, f"{vid_name}_uncomp.avi")
        if not os.path.exists(orig_vid_path):
            continue 

        # Prepare target directory (e.g., ./kth_split/train/boxing)
        out_dir = os.path.join(target_dir, split, action)
        os.makedirs(out_dir, exist_ok=True)

        # Read and Split Video
        cap = cv2.VideoCapture(orig_vid_path)
        fps = cap.get(cv2.CAP_PROP_FPS) or 25.0
        w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        fourcc = cv2.VideoWriter_fourcc(*'XVID')

        for i, (start_f, end_f) in enumerate(ranges):
            start_f, end_f = int(start_f), int(end_f)
            out_path = os.path.join(out_dir, f"{vid_name}_chunk{i}.avi")

            # Skip if file was already created in a previous notebook run
            if os.path.exists(out_path): 
                continue

            out = cv2.VideoWriter(out_path, fourcc, fps, (w, h))

            # Seek to start frame and read until end frame
            cap.set(cv2.CAP_PROP_POS_FRAMES, start_f - 1)
            for _ in range(start_f, end_f + 1):
                ret, frame = cap.read()
                if not ret: break
                out.write(frame)

            out.release()

        cap.release()

    print("Partitioning complete!")
    return target_dir
```

## Helper Functions

```python
def _find_classes(dir):
    # Only iterate over directories to ignore files like '00sequences.txt'
    classes = [d.name for d in os.scandir(dir) if d.is_dir()]
    classes.sort()
    class_to_idx = {cls_name: i for i, cls_name in enumerate(classes)}
    return classes, class_to_idx

def get_samples(root, extensions=(".mp4", ".avi")):
    _, class_to_idx = _find_classes(root)
    return make_dataset(root, class_to_idx, extensions=extensions)
```

## Iterable Dataset Definition

```python
class KTHVideoDataset(torch.utils.data.Dataset):
    def __init__(self, root, transform=None, clip_len=64, train=True, cache_dir=None):
        super().__init__()
        all_samples = get_samples(root)
        self.clip_len = clip_len
        self.transform = transform
        self.train = train  # True → random clip start; False → center clip start

        # Decode-once cache: decoded frames are written to .npy in cache_dir on
        # first access, then reloaded (no cv2 decode) on every later epoch.
        # This removes the dominant CPU bottleneck (re-decoding .avi each epoch).
        # The cache is shared across all DataLoader workers via the filesystem.
        self.cache_dir = cache_dir
        if self.cache_dir is not None:
            os.makedirs(self.cache_dir, exist_ok=True)

        # Keep every readable video. Clips shorter than clip_len are NOT
        # discarded — they are looped up to clip_len in __getitem__. This is
        # important for KTH: many running/jogging chunks are < 64 frames, and
        # dropping them would gut those classes (and empty the val set).
        # We only filter out videos that can't be read at all (0 frames).
        self.samples = []
        for path, target in all_samples:
            cap = cv2.VideoCapture(path)
            total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            cap.release()
            if total_frames >= 1:
                self.samples.append((path, target))
            else:
                print(f"Warning: skipping unreadable video (0 frames): {path}")

    def __len__(self):
        return len(self.samples)

    def _decode_video(self, path):
        """Decode all frames of a video into a (T, H, W, 3) uint8 array (RGB)."""
        cap = cv2.VideoCapture(path)
        if not cap.isOpened():
            raise RuntimeError(f"Could not open video: {path}")
        frames = []
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            frames.append(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        cap.release()
        if not frames:
            raise RuntimeError(f"No frames could be read from: {path}")
        return np.stack(frames, axis=0)

    def _load_frames(self, path):
        """Return all frames as a (T, H, W, 3) uint8 array, using the disk cache."""
        if self.cache_dir is None:
            return self._decode_video(path)

        # Cache filename from the video's (unique) basename, e.g.
        # person01_boxing_d1_chunk0.avi -> person01_boxing_d1_chunk0.npy
        key = os.path.splitext(os.path.basename(path))[0]
        cache_path = os.path.join(self.cache_dir, key + ".npy")

        if os.path.exists(cache_path):
            return np.load(cache_path)

        frames = self._decode_video(path)
        # Atomic write (tmp + rename) so concurrent workers never read a
        # half-written file. The tmp name MUST end in .npy, otherwise np.save
        # appends .npy itself and os.replace can't find the file it wrote.
        tmp_path = os.path.join(self.cache_dir, f"{key}.{os.getpid()}.tmp.npy")
        np.save(tmp_path, frames)
        os.replace(tmp_path, cache_path)
        return frames

    def __getitem__(self, idx):
        path, target = self.samples[idx]

        frames = self._load_frames(path)  # (T, H, W, 3) uint8
        total_frames = frames.shape[0]

        if total_frames >= self.clip_len:
            # Enough frames: take a contiguous temporal crop.
            # Training: random start for augmentation. Eval: deterministic center.
            if self.train:
                start = random.randint(0, total_frames - self.clip_len)
            else:
                start = (total_frames - self.clip_len) // 2
            indices = [start + i for i in range(self.clip_len)]
        else:
            # Fewer frames than clip_len: loop the clip cyclically up to clip_len.
            # Periodic actions (running/jogging/walking) tolerate this well.
            # Training: random phase offset adds variety. Eval: start at 0.
            if self.train:
                offset = random.randint(0, total_frames - 1)
            else:
                offset = 0
            indices = [(offset + i) % total_frames for i in range(self.clip_len)]

        # Fancy indexing returns a contiguous (clip_len, H, W, 3) uint8 copy.
        clip = frames[indices]
        # -> (clip_len, 3, H, W) float in [0, 1]
        video = torch.from_numpy(clip).permute(0, 3, 1, 2).float() / 255.0

        if self.transform:
            video = self.transform(video)

        return video, target
```

## Model Setup


```python
def setup_model(num_classes, freeze=False):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    weights = S3D_Weights.KINETICS400_V1
    model = s3d(weights=weights)

    if freeze:
        # Freeze all pretrained weights
        for param in model.parameters():
            param.requires_grad = False

    # Replace only the classifier head
    in_channels = model.classifier[1].in_channels
    model.classifier[1] = nn.Conv3d(in_channels, num_classes, kernel_size=1)

    model = model.to(device)

    return model, device
```

```python
def model_flops(clip_len=64):
    # Dynamically get the device from the (already set-up) model
    device = next(model.parameters()).device

    # Match the tensor the model actually receives: (B, C, T, H, W) with
    # T = clip_len frames (so FLOPs scale with the real temporal length).
    inputs = torch.randn(1, 3, clip_len, 224, 224).to(device)

    # Return the (lazy) analysis object; wrap with flop_count_table() to print
    return FlopCountAnalysis(model, inputs)
```

## Training Model Function

```python
def train_model(model, dataloaders, device, class_names, save_dir, num_epochs=30,
                learning_rate=1e-2, momentum=0.9, weight_decay=1e-5,
                lr_step_size=10, lr_gamma=0.1, freeze=False, unfreeze_epoch=None,
                use_amp=True):
    criterion = nn.CrossEntropyLoss()

    # Paper-faithful optimizer: synchronous SGD with momentum 0.9.
    # SGD + momentum tends to generalize better than Adam when fine-tuning
    # pretrained 3D CNNs on small video datasets like KTH.
    optimizer = optim.SGD(model.parameters(), lr=learning_rate,
                          momentum=momentum, weight_decay=weight_decay)

    # Step-decay LR schedule (the paper decays the LR by 10x at staged steps:
    # 0.1 -> 0.01 -> 0.001). StepLR reproduces this staged decay.
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=lr_step_size, gamma=lr_gamma)

    # A100 mixed precision: bfloat16 autocast. bf16 has the dynamic range of
    # fp32, so (unlike fp16) it needs no GradScaler and is numerically stable.
    amp_enabled = use_amp and device.type == "cuda"

    # Initialize TorchMetrics for accuracy tracking
    num_classes = len(class_names)
    acc_metric = MulticlassAccuracy(num_classes=num_classes, average='micro').to(device)

    # Dictionary to store metrics for plotting
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': [], 'lr': []}

    best_val_acc = 0.0
    best_model_path = os.path.join(save_dir, "best_model.pth")
    
    print(f"Starting Training Loop... (AMP/bf16: {amp_enabled})")
    for epoch in range(num_epochs):

        # Unfreezing backbone in case it was frozen.
        # Rebuild the optimizer + scheduler so the full network is fine-tuned
        # from a lower (10x reduced) base LR, preserving the pretrained
        # temporal representations, then continue the staged step decay.
        if freeze and unfreeze_epoch is not None and epoch == unfreeze_epoch:
            print("Unfreezing full backbone...")
            for param in model.parameters():
                param.requires_grad = True
            optimizer = optim.SGD(model.parameters(), lr=learning_rate * lr_gamma,
                                  momentum=momentum, weight_decay=weight_decay)
            scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=lr_step_size, gamma=lr_gamma)

        current_lr = optimizer.param_groups[0]['lr']
        print(f"\nEpoch {epoch + 1}/{num_epochs} (lr: {current_lr:.2e})")
        print("-" * 15)

        # Each epoch has a training and validation phase
        for phase in ['train', 'validation']:
            if phase == 'train':
                model.train()  # Set model to training mode
            else:
                model.eval()   # Set model to evaluate mode

            running_loss = 0.0
            batch_count = 0
            
            # Iterate over data
            for inputs, labels in dataloaders[phase]:
                # non_blocking=True overlaps the host->device copy with compute
                # (works because the DataLoader uses pinned memory).
                inputs = inputs.to(device, non_blocking=True)
                labels = labels.to(device, non_blocking=True)
                
                optimizer.zero_grad(set_to_none=True)
                
                # Forward pass: track history only if in train phase
                with torch.set_grad_enabled(phase == 'train'):
                    with torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=amp_enabled):
                        outputs = model(inputs)
                        loss = criterion(outputs, labels)
                    _, preds = torch.max(outputs, 1)
                    
                    # Backward pass & optimize only if in training phase.
                    # No GradScaler needed with bf16.
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()
                        
                running_loss += loss.item()
                acc_metric.update(preds, labels)
                batch_count += 1
                
            # Calculate epoch-level metrics
            epoch_loss = running_loss / batch_count
            epoch_acc = acc_metric.compute().item()
            acc_metric.reset() # Reset metric for the next phase/epoch
            
            print(f"{phase.capitalize()} Loss: {epoch_loss:.4f} | Acc: {epoch_acc * 100:.2f}%")
            
            # Save metrics to history
            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc)
                history['lr'].append(current_lr)
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc)

                # --- SAVE BEST MODEL WEIGHTS ---
                if epoch_acc > best_val_acc:
                    best_val_acc = epoch_acc
                    torch.save(model.state_dict(), best_model_path)
                    print(f" New best model saved (val acc: {best_val_acc * 100:.2f}%)")

        # Step the LR scheduler once per epoch (after both phases)
        scheduler.step()

    # --- LOAD BEST MODEL WEIGHTS ---
    print(f"\nRestoring best model weights (val acc: {best_val_acc * 100:.2f}%)")
    model.load_state_dict(torch.load(best_model_path))

    return model, history
```

## Model Evaluation Function

```python
def evaluate_model(model, dataloader, device, class_names, save_dir, use_amp=True):
    model.eval()
    num_classes = len(class_names)
    amp_enabled = use_amp and device.type == "cuda"
    
    accuracy_metric = MulticlassAccuracy(num_classes=num_classes).to(device)
    confmat_metric = MulticlassConfusionMatrix(num_classes=num_classes).to(device)

    all_preds = []
    all_labels = []

    print("\nEvaluating model with TorchMetrics...")
    # inference_mode is a faster, stricter no_grad for pure inference.
    with torch.inference_mode():
        for inputs, labels in dataloader:
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)
            
            with torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=amp_enabled):
                outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            accuracy_metric.update(preds, labels)
            confmat_metric.update(preds, labels)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    final_accuracy = accuracy_metric.compute()
    cm_tensor = confmat_metric.compute()

    print(f"Overall Accuracy: {final_accuracy.item() * 100:.2f}%")
    
    # --- Sklearn Classification Report ---
    print("Classification Report:")
    report = classification_report(all_labels, all_preds, target_names=class_names)
    print(report)
    
    # Save the text report to your run directory
    report_path = os.path.join(save_dir, "classification_report.txt")
    with open(report_path, "w") as f:
        f.write(f"Overall Accuracy: {final_accuracy.item() * 100:.2f}%\n\n")
        f.write(report)
    print(f"Classification report saved to {report_path}")
    
    # --- Plotting the Confusion Matrix ---
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm_tensor.cpu().numpy(), annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('True Label', fontsize=12)
    plt.title('Confusion Matrix', fontsize=14)
    # Save the confusion matrix
    plt.savefig(os.path.join(save_dir, "confusion_matrix.png"), dpi=300, bbox_inches='tight')
    plt.show()

    accuracy_metric.reset()
    confmat_metric.reset()
```

## Visualizing Results Function

```python
def visualize_predictions(model, dataset, device, class_names, save_dir, num_images=4):
    model.eval()

    # Randomly sample indices from the full dataset
    indices = random.sample(range(len(dataset)), num_images)

    inputs = torch.stack([dataset[i][0] for i in indices])
    labels = torch.tensor([dataset[i][1] for i in indices])

    with torch.no_grad():
        outputs = model(inputs.to(device))
        _, preds = torch.max(outputs, 1)

    preds = preds.cpu()

    mean = torch.tensor([0.43216, 0.394666, 0.37645]).view(3, 1, 1)
    std = torch.tensor([0.22803, 0.22145, 0.216989]).view(3, 1, 1)

    fig, axes = plt.subplots(1, num_images, figsize=(16, 4))
    if num_images == 1:
        axes = [axes]

    for i in range(num_images):
        T = inputs.size(2)
        mid_frame = inputs[i, :, T//2, :, :]
        mid_frame = mid_frame * std + mean
        mid_frame = torch.clamp(mid_frame, 0, 1)

        true_label = class_names[labels[i].item()]
        pred_label = class_names[preds[i].item()]

        ax = axes[i]
        ax.imshow(mid_frame.permute(1, 2, 0).numpy())
        color = "green" if true_label == pred_label else "red"
        ax.set_title(f"True: {true_label}\nPred: {pred_label}", color=color, fontweight='bold')
        ax.axis('off')

    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, "predictions_sample.png"), dpi=300, bbox_inches='tight')
    plt.show()
```

## Plotting Learning Curves

```python
def plot_learning_curves(history, save_dir):
    epochs = range(1, len(history['train_loss']) + 1)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Plot Loss
    ax1.plot(epochs, history['train_loss'], 'b-', label='Training Loss', marker='o')
    ax1.plot(epochs, history['val_loss'], 'r-', label='Validation Loss', marker='o')
    ax1.set_title('Training and Validation Loss', fontsize=14)
    ax1.set_xlabel('Epochs', fontsize=12)
    ax1.set_ylabel('Loss', fontsize=12)
    ax1.legend()
    ax1.grid(True, linestyle='--', alpha=0.7)
    
    # Plot Accuracy
    ax2.plot(epochs, history['train_acc'], 'b-', label='Training Accuracy', marker='o')
    ax2.plot(epochs, history['val_acc'], 'r-', label='Validation Accuracy', marker='o')
    ax2.set_title('Training and Validation Accuracy', fontsize=14)
    ax2.set_xlabel('Epochs', fontsize=12)
    ax2.set_ylabel('Accuracy', fontsize=12)
    ax2.legend()
    ax2.grid(True, linestyle='--', alpha=0.7)
    
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, "learning_curves.png"), dpi=300, bbox_inches='tight')
    plt.show()
```

```python
# Hyperparameters
SEED = 0
TEST_SPLIT = 0.2
VAL_SPLIT = 0.2
BATCH_SIZE = 8
NR_EPOCHS = 30
CLIP_LEN = 64           # paper uses 64 frames for all training/eval (was 32)

# Optimizer: paper-faithful synchronous SGD with momentum + staged step-decay LR
LEARNING_RATE = 1e-2    # SGD base LR (was Adam 1e-3)
MOMENTUM = 0.9
WEIGHT_DECAY = 1e-5
LR_STEP_SIZE = 10       # decay LR by LR_GAMMA every LR_STEP_SIZE epochs
LR_GAMMA = 0.1          # 10x staged decay, mirroring 0.1 -> 0.01 -> 0.001

FREEZE = True
UNFREEZE_EPOCH = 15

# Performance toggles (A100)
USE_AMP = True          # bfloat16 mixed precision (faster, ~half memory)
DETERMINISTIC = False   # False -> cuDNN autotuner on (faster, non-reproducible)

# DataLoader memory/throughput knobs.
# Staged (pinned) RAM ~= NUM_WORKERS * PREFETCH_FACTOR * BATCH_SIZE * clip_bytes.
# A 64-frame 224x224 float clip is ~38.5 MB, so the OLD 16 x 4 x 8 setting staged
# ~20 GB of pinned RAM and the OS OOM-killer killed the kernel. 8 x 2 x 8 ~= 5 GB.
NUM_WORKERS = 8
PREFETCH_FACTOR = 2

# Decode-once frame cache. Use the server's fast scratch dir ($TMPDIR) if set,
# otherwise fall back to a local ./cache folder. Decoded frames are stored as
# .npy here so videos are decoded only once instead of every epoch.
CACHE_DIR = os.path.join(os.environ.get("TMPDIR", "./cache"), "kth_frame_cache")

# Nvidia A100 GPU optimization
dataloader_kwargs = {
    'num_workers': NUM_WORKERS,
    'pin_memory': True,   # faster CPU→GPU transfers on A100
    'persistent_workers': True,  # keeps workers alive between epochs
    'prefetch_factor': PREFETCH_FACTOR,  # batches each worker preloads ahead
}

# Setting seed
set_seed(SEED, deterministic=DETERMINISTIC)

# --- CREATE A RUN DIRECTORY ---
# Creates a folder like: ./runs/run_20231024_153022
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
run_dir = f"./runs/s3d_v2_{timestamp}"
os.makedirs(run_dir, exist_ok=True)
print(f"Saving all results to: {run_dir}")
print(f"Frame cache dir: {CACHE_DIR}")

# --- SAVE HYPERPARAMETERS ---
hyperparams = {
    "SEED": SEED,
    "TEST_SPLIT": TEST_SPLIT,
    "VAL_SPLIT": VAL_SPLIT,
    "BATCH_SIZE": BATCH_SIZE,
    "NR_EPOCHS": NR_EPOCHS,
    "CLIP_LEN": CLIP_LEN,
    "OPTIMIZER": "SGD",
    "LEARNING_RATE": LEARNING_RATE,
    "MOMENTUM": MOMENTUM,
    "WEIGHT_DECAY": WEIGHT_DECAY,
    "LR_STEP_SIZE": LR_STEP_SIZE,
    "LR_GAMMA": LR_GAMMA,
    "FREEZE": FREEZE,
    "UNFREEZE_EPOCH": UNFREEZE_EPOCH if FREEZE else None,
    "USE_AMP": USE_AMP,
    "DETERMINISTIC": DETERMINISTIC,
    "NUM_WORKERS": NUM_WORKERS,
    "PREFETCH_FACTOR": PREFETCH_FACTOR,
    "MODEL": "S3D (Kinetics400 Pre-trained)"
}
with open(os.path.join(run_dir, "hyperparams.json"), "w") as f:
    json.dump(hyperparams, f, indent=4)

# --- DATA & MODEL SETUP ---

# Class labels specific to KTH
class_names = ['boxing', 'handclapping', 'handwaving', 'jogging', 'running', 'walking']

# Randomly Partitions Data (Creates ./kth_split if it doesn't exist)
split_dir = partition_kth_dataset(orig_dir=data_path, target_dir="./kth_split", test_split=TEST_SPLIT, val_split=VAL_SPLIT, seed=SEED)

# Setup Model
print("\nSetting up model...")
model, device = setup_model(len(class_names), freeze=FREEZE)

# Printing Model Characteristics
print("\nModel Characteristics...")
flops = model_flops(CLIP_LEN)
print(flop_count_table(flops))

# --- SAVE MODEL CHARACTERISTICS (params + FLOPs) ---
# FlopCountAnalysis counts multiply-accumulates for an input matching the real
# training tensor: shape (1, 3, CLIP_LEN, 224, 224). total() matches the table's
# top "model" row.
FLOPS_INPUT_SHAPE = [1, 3, CLIP_LEN, 224, 224]
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
total_flops = flops.total()
model_characteristics = {
    "model": "S3D (Kinetics400 Pre-trained)",
    "flops_input_shape": FLOPS_INPUT_SHAPE,
    "total_params": total_params,
    "total_params_M": round(total_params / 1e6, 4),
    "trainable_params": trainable_params,
    "trainable_params_M": round(trainable_params / 1e6, 4),
    "total_flops": total_flops,
    "total_gflops": round(total_flops / 1e9, 4),
}
with open(os.path.join(run_dir, "model_characteristics.json"), "w") as f:
    json.dump(model_characteristics, f, indent=4)
print(f"Saved model characteristics -> "
      f"{model_characteristics['total_params_M']}M params "
      f"({model_characteristics['trainable_params_M']}M trainable), "
      f"{model_characteristics['total_gflops']}G FLOPs")

# Get the transforms
weights = S3D_Weights.KINETICS400_V1
preprocess = weights.transforms()

# Define paths for all three splits
train_dir = os.path.join(split_dir, "train")
val_dir = os.path.join(split_dir, "validation")
test_dir = os.path.join(split_dir, "test")

# Set up DataLoaders for Train and Validation (used during the training loop)
dataset_train = KTHVideoDataset(root=train_dir, transform=preprocess, clip_len=CLIP_LEN, train=True,  cache_dir=CACHE_DIR)
dataset_val   = KTHVideoDataset(root=val_dir,   transform=preprocess, clip_len=CLIP_LEN, train=False, cache_dir=CACHE_DIR)

# Pack them into a dictionary for the training function
dataloaders = {
    'train': DataLoader(dataset_train, batch_size=BATCH_SIZE, shuffle=True, **dataloader_kwargs),
    'validation': DataLoader(dataset_val, batch_size=BATCH_SIZE, shuffle=False, **dataloader_kwargs)
}

# --- TRAIN & SAVE MODEL ---
model, history = train_model(
    model=model, 
    dataloaders=dataloaders, 
    device=device, 
    class_names=class_names,
    save_dir=run_dir,
    num_epochs=NR_EPOCHS, 
    learning_rate=LEARNING_RATE,
    momentum=MOMENTUM,
    weight_decay=WEIGHT_DECAY,
    lr_step_size=LR_STEP_SIZE,
    lr_gamma=LR_GAMMA,
    freeze=FREEZE,
    unfreeze_epoch=UNFREEZE_EPOCH,
    use_amp=USE_AMP
)

# Plot and save the Learning Curves
plot_learning_curves(history, save_dir=run_dir)

# --- EVALUATE & VISUALIZE ---
print("\nRunning Final Evaluation on Test Set...")

#Initialize Testing Datasets & Loaders
dataset_test  = KTHVideoDataset(root=test_dir,  transform=preprocess, clip_len=CLIP_LEN, train=False, cache_dir=CACHE_DIR)
dataloader_test = DataLoader(dataset_test, batch_size=BATCH_SIZE, **dataloader_kwargs) 

#Evaluate the Model
evaluate_model(model, dataloader_test, device, class_names, save_dir=run_dir, use_amp=USE_AMP)

# Visualize Predictions
visualize_predictions(model, dataloader_test.dataset, device, class_names, save_dir=run_dir)
```
