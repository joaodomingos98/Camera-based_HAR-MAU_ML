## Split videos into video clips and divide in train, val, test

```python
from PIL import Image

def partition_kth_dataset(orig_dir, target_dir="./kth_split", test_split=0.2, val_split=0.2, seed=0, clip_len=16, sampling="uniform", jpg_quality=95):
    """
    Reads 00sequences.txt, extracts active frame clips, and saves them as
    JPEG frames organized into train/validation/test folders by subject ID.

    Parameters:
    - orig_dir:     Path to original KTH dataset
    - target_dir:   Where to save the partitioned frames
    - test_split:   Fraction of subjects for test (default 0.2)
    - val_split:    Fraction of remaining subjects for val (default 0.2)
    - seed:         Random seed for reproducibility
    - clip_len:     Number of frames to extract per clip
    - sampling:     "uniform" → evenly distributed, "clip" → center clip
    - jpg_quality:  JPEG compression quality (default 95)
    """
    if os.path.exists(target_dir) and os.path.exists(os.path.join(target_dir, "train")):
        print(f"Dataset already partitioned at '{target_dir}'. Skipping.")
        print("Delete the folder first if you want to re-roll splits.")
        return target_dir

    random.seed(seed)

    all_subjects = [f"{i:02d}" for i in range(1, 26)]
    random.shuffle(all_subjects)

    num_total     = len(all_subjects)
    num_test      = int(num_total * test_split)
    num_train_val = num_total - num_test
    num_val       = int(num_train_val * val_split)

    test_p  = all_subjects[:num_test]
    val_p   = all_subjects[num_test : num_test + num_val]
    train_p = all_subjects[num_test + num_val :]

    print(f"Subject splits → Train: {len(train_p)}, Val: {len(val_p)}, Test: {len(test_p)}")

    os.makedirs(target_dir, exist_ok=True)

    with open(os.path.join(target_dir, "split_records.json"), "w") as f:
        json.dump({"train": train_p, "validation": val_p, "test": test_p}, f, indent=4)

    action_to_idx = {"boxing": 0, "handclapping": 1, "handwaving": 2,
                     "jogging": 3, "running": 4, "walking": 5}
    manifests = {"train": [], "validation": [], "test": []}

    seq_file = os.path.join(orig_dir, "00sequences.txt")
    if not os.path.exists(seq_file):
        print(f"Error: could not find 00sequences.txt in {orig_dir}")
        return orig_dir

    with open(seq_file, 'r') as f:
        lines = f.readlines()

    for line in lines:
        line = line.strip()
        if not line or "frames" not in line or "*missing*" in line:
            continue

        parts    = line.split()
        vid_name = parts[0]
        ranges   = re.findall(r'(\d+)-(\d+)', " ".join(parts[2:]))

        person_id = vid_name.split('_')[0].replace('person', '')
        action    = vid_name.split('_')[1]

        if   person_id in train_p: split = 'train'
        elif person_id in val_p:   split = 'validation'
        elif person_id in test_p:  split = 'test'
        else: continue

        if action not in action_to_idx:
            print(f"Warning: unknown action '{action}', skipping.")
            continue

        orig_vid_path = os.path.join(orig_dir, action, f"{vid_name}_uncomp.avi")
        if not os.path.exists(orig_vid_path):
            continue

        out_dir = os.path.join(target_dir, split, action)
        os.makedirs(out_dir, exist_ok=True)

        cap = cv2.VideoCapture(orig_vid_path)

        for i, (start_f, end_f) in enumerate(ranges):
            start_f, end_f = int(start_f), int(end_f)
            clip_total = end_f - start_f + 1

            frame_dir = os.path.join(out_dir, f"{vid_name}_chunk{i}")

            # If already processed, just re-add to manifest and skip
            if os.path.exists(frame_dir):
                frame_paths = [
                    os.path.join(frame_dir, f"frame_{f_idx:04d}.jpg")
                    for f_idx in range(clip_len)
                ]
                if all(os.path.exists(fp) for fp in frame_paths):
                    manifests[split].append((frame_paths, action_to_idx[action]))
                    continue

            if clip_total < clip_len:
                print(f"Warning: skipping short clip ({clip_total} frames): {vid_name} chunk {i}")
                continue

            # Compute frame indices (00sequences.txt is 1-indexed)
            if sampling == "uniform":
                relative = torch.linspace(0, clip_total - 1, clip_len).long().tolist()
            else:  # center clip
                start_offset = (clip_total - clip_len) // 2
                relative = list(range(start_offset, start_offset + clip_len))

            frame_indices = [start_f - 1 + r for r in relative]

            # Extract frames
            frames = []
            for fi in frame_indices:
                cap.set(cv2.CAP_PROP_POS_FRAMES, fi)
                ret, frame = cap.read()
                if not ret:
                    print(f"Warning: failed to read frame {fi} from {vid_name}, skipping clip.")
                    frames = []
                    break
                frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                frames.append(frame)  # keep as numpy array for PIL saving

            if len(frames) < clip_len:
                continue

            # Save frames as JPEGs
            os.makedirs(frame_dir, exist_ok=True)
            frame_paths = []
            for f_idx, frame in enumerate(frames):
                img_path = os.path.join(frame_dir, f"frame_{f_idx:04d}.jpg")
                cv2.imwrite(img_path, frame, [cv2.IMWRITE_JPEG_QUALITY, jpg_quality])
                frame_paths.append(img_path)

            manifests[split].append((frame_paths, action_to_idx[action]))

        cap.release()

    # Save one manifest per split
    for split, entries in manifests.items():
        manifest_path = os.path.join(target_dir, split, "manifest.pt")
        torch.save(entries, manifest_path)
        print(f"{split}: {len(entries)} clips saved")

    print("Partitioning complete!")
    return target_dir
```

## The dataset class

```python
from PIL import Image
from pathlib import Path

class KTHFrameDataset(torch.utils.data.Dataset):
    def __init__(self, manifest, transform=None):
        if isinstance(manifest, (str, Path)):
            self.samples = torch.load(manifest)
        else:
            self.samples = manifest
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        frame_paths, target = self.samples[idx]  # frame_paths is now a list of jpg paths

        frames = []
        for fp in frame_paths:
            img = Image.open(fp).convert('RGB')
            frames.append(torch.tensor(
                list(img.getdata()), dtype=torch.float32
            ).reshape(img.height, img.width, 3).permute(2, 0, 1))

        video = torch.stack(frames, dim=0).div(255.0)  # (T, C, H, W)

        if self.transform:
            video = self.transform(video)

        return video, target
```

## video vision transformer implementation

The MViT apparently already have a patch embedder inside of it, so for now the S3D is not used

```python
import torch
import torch.nn as nn
import torchvision.models.video as video_models

def get_mvit_model(num_classes=6, unfreeze_blocks=0):
    """
    Parameters:
    - num_classes:     Number of output classes
    - unfreeze_blocks: How many transformer blocks from the end to unfreeze
                       0 → head only
                       1 → head + last block
                       2 → head + last 2 blocks
                      -1 → unfreeze everything (full fine-tune)
    """
    model = video_models.mvit_v2_s(weights=video_models.MViT_V2_S_Weights.KINETICS400_V1)

    # Freeze everything first
    for param in model.parameters():
        param.requires_grad = False

    # Swap classification head (automatically requires_grad=True)
    num_features = model.head[1].in_features
    model.head[1] = nn.Linear(num_features, num_classes)

    # Unfreeze the last N transformer blocks
    if unfreeze_blocks == -1:
        for param in model.parameters():
            param.requires_grad = True
    elif unfreeze_blocks > 0:
        for param in model.blocks[-unfreeze_blocks:].parameters():
            param.requires_grad = True

    # Print a summary of what's trainable
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total     = sum(p.numel() for p in model.parameters())
    print(f"Trainable params: {trainable:,} / {total:,} ({100 * trainable / total:.1f}%)")

    return model
```

## Setting up model, loss_fn and optimizer (ADAMW)

```python
import torch.optim as optim

# Move model to the right device (GPU if available, else CPU)
model = model.to(device)

# 1. Loss Function
criterion = nn.CrossEntropyLoss()

# 2. Optimizer 
# We only want to optimize parameters that require a gradient 
# (This is crucial if you froze the backbone earlier!)
optimizer = optim.AdamW([
    {"params": model.head.parameters(),           "lr": 1e-3},
    {"params": model.blocks[-1:].parameters(),    "lr": 1e-4},
], weight_decay=1e-4)

# 3. Learning Rate Scheduler (Optional but recommended)
# Reduces the learning rate if the validation loss plateaus
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', patience=2, factor=0.5)
```

### Creating training loop

```python
def train_epoch(model, dataloader, criterion, optimizer, device):
    model.train() # Set model to training mode
    
    running_loss = 0.0
    correct_preds = 0
    total_samples = 0
    
    for video_batch, labels in dataloader:
        # Move data to device and fix permutation from (B, T, C, H, W) -> (B, C, T, H, W)
        # video_batch = video_batch.permute(0, 2, 1, 3, 4)
        video_batch = video_batch.to(device)
        labels = labels.to(device)
        
        # Zero the parameter gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(video_batch)
        loss = criterion(outputs, labels)
        
        # Backward pass and optimize
        loss.backward()
        optimizer.step()
        
        # Track statistics
        running_loss += loss.item() * video_batch.size(0)
        _, preds = torch.max(outputs, 1)
        correct_preds += torch.sum(preds == labels.data)
        total_samples += video_batch.size(0)
        
    epoch_loss = running_loss / total_samples
    # epoch_acc = correct_preds.double() / total_samples 
    epoch_acc = correct_preds.float() / total_samples
    
    return epoch_loss, epoch_acc.item()
```

### Creating test loop

```python
def validate_epoch(model, dataloader, criterion, device):
    model.eval() # Set model to evaluation mode
    
    running_loss = 0.0
    correct_preds = 0
    total_samples = 0
    
    with torch.no_grad(): # Disable gradient calculation for speed and memory efficiency
        for video_batch, labels in dataloader:
            #video_batch = video_batch.permute(0, 2, 1, 3, 4)
            video_batch = video_batch.to(device)
            labels = labels.to(device)
            
            outputs = model(video_batch)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item() * video_batch.size(0)
            _, preds = torch.max(outputs, 1)
            correct_preds += torch.sum(preds == labels.data)
            total_samples += video_batch.size(0)
            
    epoch_loss = running_loss / total_samples
    # epoch_acc = correct_preds.double() / total_samples
    epoch_acc = correct_preds.float() / total_samples
    
    return epoch_loss, epoch_acc.item()
```

## Running training and validation

```python
# History tracking
history = {
    "train_loss": [],
    "train_acc":  [],
    "val_loss":   [],
    "val_acc":    [],
}

num_epochs = 30
best_val_loss = float('inf')

print(f"Starting training on device: {device}\n" + "-"*30)

for epoch in range(num_epochs):
    # 1. Run Train and Validation phases
    train_loss, train_acc = train_epoch(model, dataloaders["train"], criterion, optimizer, device)
    val_loss, val_acc = validate_epoch(model, dataloaders["validation"], criterion, device)

    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    
    # 2. Step the learning rate scheduler based on validation loss
    scheduler.step(val_loss)

    # 3. Print Epoch Progress
    print(f"Epoch {epoch+1}/{num_epochs}")
    print(f"  Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}%")
    print(f"  Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc*100:.2f}%")
    for i, pg in enumerate(optimizer.param_groups):
        print(f"Group {i} LR: {pg['lr']:.2e}")
    
    # 4. Save Best Model checkpoint
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        torch.save(model.state_dict(), 'best_mvit_kth_model.pth')
        print("New best validation loss! Model saved.")
        
    print("-" * 30)
    
print("Training Complete!")
```

## Evaluating the model

```python
import matplotlib.pyplot as plt
import seaborn as sns

def evaluate_model(model, dataloader, device, class_names, save_dir):
    model.eval()
    num_classes = len(class_names)
    
    accuracy_metric = MulticlassAccuracy(num_classes=num_classes).to(device)
    confmat_metric = MulticlassConfusionMatrix(num_classes=num_classes).to(device)

    all_preds = []
    all_labels = []

    print("\nEvaluating model with TorchMetrics...")
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            accuracy_metric.update(preds, labels)
            confmat_metric.update(preds, labels)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    final_accuracy = accuracy_metric.compute()
    cm_tensor = confmat_metric.compute()

    print(f"Overall Accuracy: {final_accuracy.item() * 100:.2f}%")
    
    # --- Sklearn Classification Report ---
    print("Classification Report:")
    report = classification_report(all_labels, all_preds, target_names=class_names)
    print(report)
    
    # Save the text report to your run directory
    report_path = os.path.join(save_dir, "classification_report.txt")
    with open(report_path, "w") as f:
        f.write(f"Overall Accuracy: {final_accuracy.item() * 100:.2f}%\n\n")
        f.write(report)
    print(f"Classification report saved to {report_path}")
    
    # --- Plotting the Confusion Matrix ---
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm_tensor.cpu().numpy(), annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('True Label', fontsize=12)
    plt.title('Confusion Matrix', fontsize=14)
    # Save the confusion matrix
    plt.savefig(os.path.join(save_dir, "confusion_matrix.png"), dpi=300, bbox_inches='tight')
    plt.show()

    accuracy_metric.reset()
    confmat_metric.reset()
```

```python
import torch
import torch.nn.functional as F

def plot_calibration_curve(model, dataloader, device, class_names, save_dir, n_bins=10):
    model.eval()
    all_confidences = []
    all_correct     = []

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            logits = model(inputs)
            probs  = F.softmax(logits, dim=1)

            confidences = probs.max(dim=1).values
            predictions = probs.argmax(dim=1)
            correct     = (predictions == labels).float()

            all_confidences.extend(confidences.cpu().tolist())
            all_correct.extend(correct.cpu().tolist())

    # Bin by confidence
    bin_edges     = [i / n_bins for i in range(n_bins + 1)]
    bin_mean_conf = []
    bin_accuracy  = []
    bin_labels    = []

    for i in range(n_bins):
        low, high = bin_edges[i], bin_edges[i + 1]
        in_bin = [(c, cor) for c, cor in zip(all_confidences, all_correct) if low <= c < high]

        if len(in_bin) == 0:
            continue

        confs, cors = zip(*in_bin)
        bin_mean_conf.append(sum(confs) / len(confs))
        bin_accuracy.append(sum(cors) / len(cors))
        bin_labels.append(f"{low:.1f}-{high:.1f}")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

    # --- Left: Reliability diagram with bars ---
    ax1.bar(bin_mean_conf, bin_accuracy, width=0.05, alpha=0.8, label='this model')
    ax1.plot([0, 1], [0, 1], linestyle='--', color='black', label='perfect calibration')
    ax1.set_xlabel("What the model claimed (avg confidence)")
    ax1.set_ylabel("What actually happened (fraction correct)")
    ax1.set_title("Reliability diagram")
    ax1.set_xlim(0.4, 1.0)
    ax1.set_ylim(0.4, 1.0)
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # --- Right: Per-bin grouped bar chart ---
    bar_width = 0.35
    x = list(range(len(bin_labels)))
    x_claimed = [i - bar_width / 2 for i in x]
    x_actual  = [i + bar_width / 2 for i in x]

    ax2.bar(x_claimed, bin_mean_conf, width=bar_width, color='orange', alpha=0.9, label='claimed (avg confidence)')
    ax2.bar(x_actual,  bin_accuracy,  width=bar_width, color='steelblue', alpha=0.9, label='actual (fraction correct)')
    ax2.set_xticks(x)
    ax2.set_xticklabels(bin_labels)
    ax2.set_xlabel("Confidence bin")
    ax2.set_ylabel("Probability")
    ax2.set_title("Per-bin comparison: claimed vs actual")
    ax2.set_ylim(0, 1.0)
    ax2.legend()
    ax2.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, "calibration_curve.png"), dpi=150, bbox_inches='tight')
    plt.show()
```
